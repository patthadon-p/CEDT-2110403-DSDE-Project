## Get different

```python
# Cast ticket_id to string for accurate comparison
pandas_df["ticket_id"] = pandas_df["ticket_id"].astype("str")
spark_df["ticket_id"] = spark_df["ticket_id"].astype("str")


# Get Columns
columns = pandas_df.columns.tolist()
columns_indicated = [col + "_x" if col != "ticket_id" else col for col in columns]

columns_mapping = dict(zip(columns_indicated, columns))


# Join and get differences
df_pandas_not_spark = (
    pandas_df.merge(
        spark_df, on="ticket_id", how="left", indicator=True, suffixes=("_x", "_y")
    )
    .query('_merge == "left_only"')
    .drop(columns=["_merge"])
)

df_spark_not_pandas = (
    spark_df.merge(
        pandas_df, on="ticket_id", how="left", indicator=True, suffixes=("_x", "_y")
    )
    .query('_merge == "left_only"')
    .drop(columns=["_merge"])
)


# Rename columns for clarity
df_pandas_not_spark = df_pandas_not_spark.loc[:, columns_indicated]
df_pandas_not_spark = df_pandas_not_spark.rename(columns=columns_mapping)

df_spark_not_pandas = df_spark_not_pandas.loc[:, columns_indicated]
df_spark_not_pandas = df_spark_not_pandas.rename(columns=columns_mapping)


# Output the number of different rows
print("Rows in Pandas but not in Spark:", df_pandas_not_spark.shape[0])
print("Rows in Spark but not in Pandas:", df_spark_not_pandas.shape[0])
```

```python
if df_pandas_not_spark.shape[0] > 0:
    display(df_pandas_not_spark.head(3))

if df_spark_not_pandas.shape[0] > 0:
    display(df_spark_not_pandas.head(3))
```

```python
# Get the raw data of the differing rows for further investigation

df_raw_pandas_not_spark = raw_df[
    raw_df["ticket_id"].isin(df_pandas_not_spark["ticket_id"])
]
df_raw_spark_not_pandas = raw_df[
    raw_df["ticket_id"].isin(df_spark_not_pandas["ticket_id"])
]
```
